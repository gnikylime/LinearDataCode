So, the Pete Dog image has pixel values spanning the entire range of possible pixel values.

First, let's convert Pete Dog to grayscale and do some linear algebra to him.

```python
I = I.astype(float)/255
bwI = 0.2989 * I[:, :, 0] + 0.5870 * I[:, :, 1] + 0.1140 * I[:, :, 2]
plt.imshow(bwI,cmap="gray")
plt.axis("off")
plt.show()
```

There is a command in the skimage package to convert images to grayscale.  Here, we are implementing the conversion as a weighted average of the different color channels.  The weights are related to theories about human vision.  Note that in order to scale the image array non-integer decimal numbers, we needed to convert it to float.

Let's look at properties of this new image.

```python
bwI.shape
```

```python
bwI.dtype
```

So, we still have 850 rows and 800 columns and entries which are floats between 0 and 1, but now there is only one channel/slice, where the numbers are encoding luminance/brightness.

Let's zoom in on a little corner (upper left 100 by 100 pixels) of the grayscale Pete Dog.

We can scalar multiply and vector add matrices and tensors just like (1D array) vectors. Namely, to scalar multiply a matrix (or tensor) M by a scalar c, we just make a new matrix (or tensor) of the same shape and size as M, called cM, where each entry is multiplied by c. Similarly, if M and A are two matrices (or tensors) of the same size, we can add them together by making a new matrix (or tensor) of the same size as M and A, called M+A, where each entry is the some of the corresponding entries in M and A. So, just like we did with sound, we are going to add noise to the Pete Dog image. First, we generate the noise.

```python
noise = np.random.randn(*bwI.shape)
plt.imshow(noise, cmap="gray")
plt.axis("off")
```

Now we add the noise. Note that if we had tried to add noise to an image that was still uint, we would have received an error.

```python
nI = bwI + noise
plt.imshow(nI, cmap="gray")
plt.axis("off")
```

We can also make other linear combinations to scale the amount of noise.

```python
a = 0.1
plt.imshow(bwI + a * noise, cmap="gray")
plt.axis("off")
```

Let's do one last thing with color. Choose three numbers between 0 and 1 for the red, green, and blue parts. We'll generate a 400x400x3 tensor with only that color.

```python
r, g, b = 0.6, 0.1, 0.9
color_block = np.stack([
    r * np.ones((400, 400)),
    g * np.ones((400, 400)),
    b * np.ones((400, 400))
], axis=-1)

plt.imshow(color_block)
plt.title("Generated Color Block")
plt.axis("off")
```
